```python
# 3. اتصال امن به Google Drive (برای ذخیره دائم پروژه‌ها)
from google.colab import drive
import os
drive.mount('/content/drive')
os.makedirs('/content/drive/MyDrive/LuminaStudio/projects', exist_ok=True)
os.makedirs('/content/drive/MyDrive/LuminaStudio/videos', exist_ok=True)
print('✅ Google Drive Mounted Successfully.')
```

```python
# 4. دریافت امن GEMINI_API_KEY از Colab Secrets
from google.colab import userdata
import os
try:
    os.environ['GEMINI_API_KEY'] = userdata.get('GEMINI_API_KEY')
    print('✅ API Key loaded successfully.')
except:
    print('❌ Error: GEMINI_API_KEY is not set in Colab Secrets.')
```
